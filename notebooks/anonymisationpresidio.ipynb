```python
from pathlib import Path
import spacy
from presidio_analyzer import AnalyzerEngine
from presidio_anonymizer import AnonymizerEngine

input_dir = Path("corpus_original")
output_dir = Path("corpus_anonymise_presidio")
output_dir.mkdir(exist_ok=True)

nlp_fr = spacy.load("fr_core_news_lg")
nlp_nl = spacy.load("nl_core_news_lg")
nlp_en = spacy.load("en_core_web_lg")

analyzer = AnalyzerEngine()
anonymizer = AnonymizerEngine()

def detect_lang(file_path):
    path = str(file_path).upper()
    if "FR" in path:
        return "fr"
    elif "NL" in path:
        return "nl"
    elif "EN" in path:
        return "en"
    return "en"

def anonymize_entities(text, lang):
    if lang == "fr":
        doc = nlp_fr(text)
    elif lang == "nl":
        doc = nlp_nl(text)
    else:
        doc = nlp_en(text)

    new_text = text

    for ent in sorted(doc.ents, key=lambda ent: ent.start_char, reverse=True):
        if ent.label_ in ["PER", "PERSON"]:
            replacement = "<PERSON>"
        elif ent.label_ in ["LOC", "GPE"]:
            replacement = "<LOCATION>"
        elif ent.label_ == "ORG":
            replacement = "<ORGANIZATION>"
        else:
            continue

        new_text = new_text[:ent.start_char] + replacement + new_text[ent.end_char:]

    return new_text

for file_path in input_dir.rglob("*.txt"):
    text = file_path.read_text(encoding="utf-8")
    lang = detect_lang(file_path)

    text = anonymize_entities(text, lang)

    results = analyzer.analyze(
        text=text,
        language="en"
    )

    anonymized = anonymizer.anonymize(
        text=text,
        analyzer_results=results
    )

    relative_path = file_path.relative_to(input_dir)
    output_file = output_dir / relative_path
    output_file.parent.mkdir(parents=True, exist_ok=True)
    output_file.write_text(anonymized.text, encoding="utf-8")

    print(f"{file_path.name} anonymisé ({lang})")
```
