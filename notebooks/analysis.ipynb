```python
nlp_en = spacy.load("en_core_web_lg")
nlp_fr = spacy.load("fr_core_news_lg")
nlp_nl = spacy.load("nl_core_news_lg")
```

```python
base_path = Path("corpus_original")

corpus_paths = {
    "HaveYourSay_EN": base_path / "HaveYourSay" / "EN",
    "HaveYourSay_FR": base_path / "HaveYourSay" / "FR",
    "HaveYourSay_NL": base_path / "HaveYourSay" / "NL",
    "Jurisprudence_FR": base_path / "Jurisprudence" / "FR",
    "Jurisprudence_NL": base_path / "Jurisprudence" / "NL",
}
```

```python
documents = []

for corpus, folder in corpus_paths.items():
    for file_path in folder.glob("*.txt"):
        documents.append({
            "corpus": corpus,
            "fichier": file_path.name,
            "texte": file_path.read_text(encoding="utf-8")
        })

df = pd.DataFrame(documents)

df
```

```python
def clean_text(text):
    text = text.replace("\n", " ")
    text = text.replace("\t", " ")
    text = " ".join(text.split())
    return text


def get_nlp_model(corpus):
    if corpus.endswith("_EN"):
        return nlp_en
    elif corpus.endswith("_FR"):
        return nlp_fr
    elif corpus.endswith("_NL"):
        return nlp_nl
```

```python
labels_utiles = ["PERSON", "PER", "ORG", "GPE", "LOC"]

resultats = []

for _, row in df.iterrows():
    nlp = get_nlp_model(row["corpus"])
    texte_nettoye = clean_text(row["texte"])
    doc = nlp(texte_nettoye)

    for ent in doc.ents:
        label = ent.label_

        if label in labels_utiles:
            if label == "PER":
                label = "PERSON"
            elif label == "GPE":
                label = "LOC"

            resultats.append({
                "corpus": row["corpus"],
                "fichier": row["fichier"],
                "entite": ent.text,
                "label": label
            })

ner_df = pd.DataFrame(resultats)

ner_df = ner_df.drop_duplicates().reset_index(drop=True)

ner_df
```

```python
pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)

display(ner_df)
```

```python
stats_labels = ner_df["label"].value_counts().reset_index()
stats_labels.columns = ["label", "nombre"]

stats_labels
```

```python
stats_par_corpus = pd.crosstab(
    ner_df["corpus"],
    ner_df["label"]
).reset_index()

stats_par_corpus
```

```python
ner_df.to_excel("tableau_ner_complet.xlsx", index=False)
stats_labels.to_excel("statistiques_labels_ner.xlsx", index=False)
stats_par_corpus.to_excel("statistiques_ner_par_corpus.xlsx", index=False)
```
