```python
from pathlib import Path
import spacy
```

```python
input_dir = Path("corpus_original")
output_dir = Path("corpus_anonymise_spacy")

output_dir.mkdir(exist_ok=True)
```

```python
nlp_fr = spacy.load("fr_core_news_lg")
nlp_nl = spacy.load("nl_core_news_lg")
nlp_en = spacy.load("en_core_web_lg")
```

```python
def detect_language(file_path):
    path = str(file_path).upper()

    if "FR" in path:
        return "fr"
    elif "NL" in path:
        return "nl"
    elif "EN" in path:
        return "en"
    else:
        return "en"
```

```python
def anonymize_text_spacy(text, lang):

    if lang == "fr":
        doc = nlp_fr(text)
    elif lang == "nl":
        doc = nlp_nl(text)
    else:
        doc = nlp_en(text)

    anonymized_text = text

    for ent in sorted(doc.ents, key=lambda e: e.start_char, reverse=True):

        if ent.label_ in ["PER", "PERSON"]:
            replacement = "<PERSON>"

        elif ent.label_ in ["LOC", "GPE"]:
            replacement = "<LOCATION>"

        elif ent.label_ == "ORG":
            replacement = "<ORGANIZATION>"

        else:
            continue

        anonymized_text = (
            anonymized_text[:ent.start_char]
            + replacement
            + anonymized_text[ent.end_char:]
        )

    return anonymized_text
```

```python
for file_path in input_dir.rglob("*.txt"):

    text = file_path.read_text(encoding="utf-8")

    lang = detect_language(file_path)

    anonymized_text = anonymize_text_spacy(text, lang)

    relative_path = file_path.relative_to(input_dir)
    output_file = output_dir / relative_path
    output_file.parent.mkdir(parents=True, exist_ok=True)

    output_file.write_text(anonymized_text, encoding="utf-8")

    print(f"{file_path.name} anonymisé avec spaCy ({lang})")
```
